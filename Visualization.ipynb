```python
# Load the cleaned dataset

df = pd.read_csv("disease_prediction_cleaned_eda.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
# Check dataset information

print("Dataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Disease distribution

plt.figure(figsize=(10, 6))

df["Disease"].value_counts().plot(kind="bar")

plt.title("Disease Distribution")
plt.xlabel("Disease")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Gender distribution

plt.figure(figsize=(7, 5))

df["Gender"].value_counts().plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Gender Distribution of Patients")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Age distribution

plt.figure(figsize=(9, 5))

plt.hist(df["Age"], bins=15, edgecolor="black")

plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

```python
# Age group distribution

plt.figure(figsize=(8, 5))

df["Age_Group"].value_counts().sort_index().plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Age Group Distribution of Patients")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Blood pressure distribution

plt.figure(figsize=(9, 5))

plt.hist(df["Blood_Pressure"], bins=15, edgecolor="black")

plt.title("Blood Pressure Distribution")
plt.xlabel("Blood Pressure")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

```python
# Sugar level distribution

plt.figure(figsize=(9, 5))

plt.hist(df["Sugar_Level"], bins=15, edgecolor="black")

plt.title("Sugar Level Distribution")
plt.xlabel("Sugar Level")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

```python
# Cholesterol distribution

plt.figure(figsize=(9, 5))

plt.hist(df["Cholesterol"], bins=15, edgecolor="black")

plt.title("Cholesterol Distribution")
plt.xlabel("Cholesterol")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

```python
# Disease vs Gender

disease_gender = pd.crosstab(
    df["Disease"],
    df["Gender"]
)

disease_gender.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Disease Distribution by Gender")
plt.xlabel("Disease")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.legend(title="Gender")
plt.tight_layout()
plt.show()
```

```python
# Disease distribution by Age Group

disease_age = pd.crosstab(
    df["Disease"],
    df["Age_Group"]
)

disease_age.plot(
    kind="bar",
    figsize=(11, 6)
)

plt.title("Disease Distribution by Age Group")
plt.xlabel("Disease")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.legend(title="Age Group")
plt.tight_layout()
plt.show()
```

```python
# Disease distribution by Medical History

disease_history = pd.crosstab(
    df["Disease"],
    df["Medical_History"]
)

disease_history.plot(
    kind="bar",
    figsize=(11, 6)
)

plt.title("Disease Distribution by Medical History")
plt.xlabel("Disease")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.legend(title="Medical History")
plt.tight_layout()
plt.show()
```

```python
# Age distribution by Disease

plt.figure(figsize=(12, 6))

df.boxplot(
    column="Age",
    by="Disease",
    figsize=(12, 6)
)

plt.title("Age Distribution by Disease")
plt.suptitle("")
plt.xlabel("Disease")
plt.ylabel("Age")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Sugar Level by Disease

plt.figure(figsize=(12, 6))

df.boxplot(
    column="Sugar_Level",
    by="Disease",
    figsize=(12, 6)
)

plt.title("Sugar Level Distribution by Disease")
plt.suptitle("")
plt.xlabel("Disease")
plt.ylabel("Sugar Level")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Medical History Distribution

plt.figure(figsize=(10, 6))

df["Medical_History"].value_counts().plot(
    kind="bar",
    edgecolor="black"
)

plt.title("Medical History Distribution")
plt.xlabel("Medical History")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Symptoms distribution

symptoms = (
    df["Symptoms"]
    .dropna()
    .str.split(",")
    .explode()
    .str.strip()
)

symptom_counts = symptoms.value_counts()

plt.figure(figsize=(12, 6))

plt.bar(symptom_counts.index, symptom_counts.values)

plt.title("Symptom Distribution")
plt.xlabel("Symptoms")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Correlation between numerical features

numerical_data = df[
    ["Age", "Blood_Pressure", "Sugar_Level", "Cholesterol"]
]

correlation = numerical_data.corr()

plt.figure(figsize=(8, 6))

plt.imshow(correlation, interpolation="nearest")

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Between Numerical Features")

plt.colorbar()

plt.tight_layout()
plt.show()
```

```python
# Average age for each disease

average_age = df.groupby("Disease")["Age"].mean().sort_values()

plt.figure(figsize=(10, 6))

plt.bar(average_age.index, average_age.values)

plt.title("Average Age by Disease")
plt.xlabel("Disease")
plt.ylabel("Average Age")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
